```python
import pandas as pd
from IPython.display import display

url = (
    "https://raw.githubusercontent.com/andr3nf/"
    "clima_municipios_brasileiros/main/clima_historico_municipal.csv"
)
df = pd.read_csv(url)

girua = df.loc[
    (df["nome_municipio"] == "Giruá") &
    (df["uf"] == "RS")
].copy()

if girua.empty:
    raise ValueError("Giruá não foi encontrada na base.")

print("TODAS AS INFORMAÇÕES DE GIRUÁ–RS")
display(girua.T.rename(columns={girua.index[0]: "Valor"}))

cidade = girua.iloc[0]

meses = [
    "jan", "fev", "mar", "abr", "mai", "jun",
    "jul", "ago", "set", "out", "nov", "dez"
]

mensal = pd.DataFrame({
    "Mês": meses,
    "Temperatura mínima (°C)": [
        cidade[f"tmin_{mes}"] for mes in meses
    ],
    "Temperatura média (°C)": [
        cidade[f"temp_media_{mes}"] for mes in meses
    ],
    "Déficit hídrico (mm)": [
        cidade[f"def_{mes}"] for mes in meses
    ]
})

print("\nDADOS MENSAIS")
display(mensal)

menor_temperatura = mensal["Temperatura mínima (°C)"].min()

print("\nMÊS DE TEMPERATURA MÍNIMA")
display(
    mensal.loc[
        mensal["Temperatura mínima (°C)"].eq(menor_temperatura),
        ["Mês", "Temperatura mínima (°C)"]
    ]
)

maior_deficit = mensal["Déficit hídrico (mm)"].max()

print("\nMÊS DE MAIOR DÉFICIT HÍDRICO")
display(
    mensal.loc[
        mensal["Déficit hídrico (mm)"].eq(maior_deficit),
        ["Mês", "Déficit hídrico (mm)"]
    ]
)

estacoes = {
    "dez": "Verão", "jan": "Verão", "fev": "Verão",
    "mar": "Outono", "abr": "Outono", "mai": "Outono",
    "jun": "Inverno", "jul": "Inverno", "ago": "Inverno",
    "set": "Primavera", "out": "Primavera", "nov": "Primavera"
}

mensal["Estação"] = mensal["Mês"].map(estacoes)

media_estacoes = (
    mensal.groupby("Estação")["Temperatura média (°C)"]
    .mean()
    .reindex(["Verão", "Outono", "Inverno", "Primavera"])
    .round(2)
)

print("\nTEMPERATURA MÉDIA POR ESTAÇÃO")
display(media_estacoes.to_frame())
```
