```python
import numpy as np

np.set_printoptions(suppress=True, precision=2)

dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98, 11700, 82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200, 95, 120.00, 2]
])

dias = np.array([
    "Segunda", "Terça", "Quarta", "Quinta",
    "Sexta", "Sábado", "Domingo"
])

indicadores = np.array([
    "Pedidos", "Faturamento", "Clientes",
    "Ticket médio", "Reclamações"
])

# Cada variável seleciona uma coluna do array.
pedidos = dados[:, 0]
faturamento = dados[:, 1]
clientes = dados[:, 2]
ticket_medio = dados[:, 3]
reclamacoes = dados[:, 4]
```

```python
# PARTE 1 — Estrutura do array
print("1.1 — Estrutura dos dados")
print("Quantidade de linhas:", dados.shape[0])
print("Quantidade de colunas:", dados.shape[1])
print("Quantidade total de valores:", dados.size)
print("Número de dimensões:", dados.ndim)
print("Tipo dos dados:", dados.dtype)

# PARTE 2 — Indexação e slicing
print("\nOrdem das colunas:", indicadores)

print("\n2.1 — Dados de sexta-feira:")
print(dados[4])

print("\n2.2 — Faturamento de todos os dias:")
print(faturamento)

print("\n2.3 — Quantidade de reclamações:")
print(reclamacoes)

print("\n2.4 — Dados de quinta-feira até domingo:")
print(dados[3:7])

print("\n2.5 — Colunas pedidos, faturamento e clientes:")
print(dados[:, :3])
```

```python
# PARTE 3 — Cálculos vetorizados com funções NumPy
total_pedidos = np.sum(pedidos)
total_faturamento = np.sum(faturamento)
total_clientes = np.sum(clientes)

media_pedidos = np.mean(pedidos)
media_faturamento = np.mean(faturamento)
media_reclamacoes = np.mean(reclamacoes)

print("3.1 — Total de pedidos:", int(total_pedidos))
print(f"3.2 — Faturamento total: R$ {total_faturamento:.2f}")
print("3.3 — Soma dos clientes diários:", int(total_clientes))
print(f"3.4 — Média diária de pedidos: {media_pedidos:.2f}")
print(f"3.5 — Média diária de faturamento: R$ {media_faturamento:.2f}")
print(f"3.6 — Média diária de reclamações: {media_reclamacoes:.2f}")

# PARTE 4 — Localização dos maiores e menores resultados
indice_maior_pedidos = np.argmax(pedidos)
indice_menor_pedidos = np.argmin(pedidos)

indice_maior_faturamento = np.argmax(faturamento)
indice_menor_faturamento = np.argmin(faturamento)

print("\n4.1 — Pedidos")
print(
    "Maior:",
    dias[indice_maior_pedidos],
    "—", int(pedidos[indice_maior_pedidos])
)
print(
    "Menor:",
    dias[indice_menor_pedidos],
    "—", int(pedidos[indice_menor_pedidos])
)

print("\n4.2 — Faturamento")
print(
    f"Maior: {dias[indice_maior_faturamento]} "
    f"— R$ {faturamento[indice_maior_faturamento]:.2f}"
)
print(
    f"Menor: {dias[indice_menor_faturamento]} "
    f"— R$ {faturamento[indice_menor_faturamento]:.2f}"
)

amplitude_faturamento = np.max(faturamento) - np.min(faturamento)

print(f"\n4.3 — Amplitude do faturamento: R$ {amplitude_faturamento:.2f}")
```

```python
# PARTE 5 — Máscaras booleanas
mascara_faturamento = faturamento > 12000
mascara_pedidos = pedidos > 150
mascara_reclamacoes = reclamacoes > 4
mascara_abaixo_media = faturamento < media_faturamento

print("5.1 — Dias com faturamento superior a R$ 12.000:")
print(dias[mascara_faturamento])

print("\n5.2 — Dias com mais de 150 pedidos:")
print(dias[mascara_pedidos])

print("\n5.3 — Dias com mais de 4 reclamações:")
print(dias[mascara_reclamacoes])

print("\n5.4 — Faturamentos dos dias com mais de 150 pedidos:")
print(faturamento[mascara_pedidos])

print("\n5.5 — Quantidade de dias com faturamento abaixo da média:")
print(np.count_nonzero(mascara_abaixo_media))

# PARTE 6 — Classificação vetorizada
classificacao = np.where(
    faturamento >= 20000,
    "Alto desempenho",
    "Desempenho normal"
)

print("\n6.1 — Dia e classificação:")
print(np.column_stack((dias, classificacao)))

quantidade_alto = np.count_nonzero(classificacao == "Alto desempenho")
quantidade_normal = np.count_nonzero(classificacao == "Desempenho normal")

print("\n6.2 — Quantidade em cada classificação:")
print("Alto desempenho:", quantidade_alto)
print("Desempenho normal:", quantidade_normal)
```

```python
# PARTE 7 — Estatísticas do faturamento
mediana_faturamento = np.median(faturamento)

# Desvio padrão populacional: considera os sete dias analisados.
desvio_faturamento = np.std(faturamento, ddof=0)

minimo_faturamento = np.min(faturamento)
maximo_faturamento = np.max(faturamento)

# Coeficiente de variação: desvio padrão dividido pela média.
cv_faturamento = desvio_faturamento / media_faturamento * 100

print(f"7 — Média: R$ {media_faturamento:.2f}")
print(f"Mediana: R$ {mediana_faturamento:.2f}")
print(f"Desvio padrão: R$ {desvio_faturamento:.2f}")
print(f"Mínimo: R$ {minimo_faturamento:.2f}")
print(f"Máximo: R$ {maximo_faturamento:.2f}")

print(
    f"\nO faturamento variou de R$ {minimo_faturamento:.2f} "
    f"a R$ {maximo_faturamento:.2f}, com amplitude de "
    f"R$ {amplitude_faturamento:.2f}. "
    f"O desvio padrão corresponde a {cv_faturamento:.2f}% da média, "
    "indicando uma variação considerável durante a semana."
)

# PARTE 8 — axis=0 calcula por coluna, considerando todos os dias.
medias_indicadores = np.mean(dados, axis=0)
maiores_indicadores = np.max(dados, axis=0)
menores_indicadores = np.min(dados, axis=0)

print("\nOrdem dos indicadores:", indicadores)

print("\n8.1 — Média de cada indicador:")
print(np.round(medias_indicadores, 2))

print("\n8.2 — Maior valor de cada indicador:")
print(maiores_indicadores)

print("\n8.3 — Menor valor de cada indicador:")
print(menores_indicadores)

# O coeficiente de variação permite comparar unidades diferentes.
desvios_indicadores = np.std(dados, axis=0, ddof=0)
cv_indicadores = desvios_indicadores / medias_indicadores * 100
indice_maior_variacao = np.argmax(cv_indicadores)

print("\n8.4 — Coeficiente de variação (%) de cada indicador:")
print(np.column_stack((indicadores, np.round(cv_indicadores, 2))))

print(
    f"\nMaior variação relativa: "
    f"{indicadores[indice_maior_variacao]}, "
    f"com CV de {cv_indicadores[indice_maior_variacao]:.2f}%."
)
```

```python
indice_maior_reclamacoes = np.argmax(reclamacoes)

print("9.1 — Melhor dia considerando o faturamento:")
print(dias[indice_maior_faturamento])

print("\n9.2 — Esse dia também teve a maior quantidade de pedidos?")
print(
    np.where(
        pedidos[indice_maior_faturamento] == np.max(pedidos),
        "Sim", "Não"
    )
)

print("\n9.3 — Esse dia também teve o maior número de reclamações?")
print(
    np.where(
        reclamacoes[indice_maior_faturamento] == np.max(reclamacoes),
        "Sim", "Não"
    )
)

# & combina as duas condições.
mascara_integrada = (pedidos > 150) & (faturamento > 20000)

print("\n9.4 — Dias com mais de 150 pedidos e faturamento acima de R$ 20.000:")
print(dias[mascara_integrada])
print("Quantidade de dias:", np.count_nonzero(mascara_integrada))
```

```python
# Correlação mede a associação entre os indicadores nos sete dias.
correlacao_pedidos_faturamento = np.corrcoef(pedidos, faturamento)[0, 1]
correlacao_pedidos_reclamacoes = np.corrcoef(pedidos, reclamacoes)[0, 1]

print(
    f"A empresa realizou {total_pedidos:.0f} pedidos e faturou "
    f"R$ {total_faturamento:.2f} durante a semana. "
    f"A média diária de faturamento foi de R$ {media_faturamento:.2f}.\n"
)

print(
    f"{dias[indice_maior_faturamento]} apresentou o maior faturamento, "
    f"com R$ {maximo_faturamento:.2f}, enquanto "
    f"{dias[indice_menor_faturamento]} teve o menor, "
    f"com R$ {minimo_faturamento:.2f}.\n"
)

print(
    f"{dias[indice_menor_faturamento]} merece atenção pelo menor resultado "
    f"de vendas. {dias[indice_maior_reclamacoes]} merece atenção operacional "
    f"por registrar {np.max(reclamacoes):.0f} reclamações.\n"
)

print(
    f"A correlação entre pedidos e faturamento foi de "
    f"{correlacao_pedidos_faturamento:.3f}, e entre pedidos e reclamações "
    f"foi de {correlacao_pedidos_reclamacoes:.3f}. "
    "Nesta semana, os dias com mais pedidos tenderam a apresentar maior "
    "faturamento e mais reclamações. Porém, sete dias de dados não são "
    "suficientes para demonstrar uma relação de causa e efeito."
)
```
