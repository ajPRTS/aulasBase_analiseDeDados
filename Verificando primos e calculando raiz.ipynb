```python
def eh_primo(numero):
    if numero < 2:
        return False

    if numero == 2:
        return True

    if numero % 2 == 0:
        return False

    # Testa os divisores ímpares, sem incluir o próprio número.
    for divisor in range(3, numero, 2):
        if numero % divisor == 0:
            return False

    return True


numero = int(input("Digite um número inteiro: "))

if eh_primo(numero):
    print(f"{numero} é primo.")
else:
    print(f"{numero} não é primo.")
```

```python
quantidade = int(input("Quantos números primos deseja apresentar? "))

if quantidade <= 0:
    print("Digite uma quantidade maior que zero.")
else:
    primos = []
    numero = 2

    while len(primos) < quantidade:
        if eh_primo(numero):
            primos.append(numero)

        numero += 1

    print(f"Os {quantidade} primeiros números primos são:")
    print(primos)
```

```python
import math

n = float(input("Digite um número para calcular a raiz quadrada: "))

if n < 0:
    print("Números negativos não possuem raiz quadrada real.")
else:
    if n == 0:
        p = 0.0
        iteracoes = 0
    else:
        # Estimativa inicial diferente de zero.
        b = n if n >= 1 else 1.0

        p = (b + (n / b)) / 2
        iteracoes = 1

        # Repete até que |n - p²| seja menor que 0,0001.
        while abs(n - p ** 2) >= 0.0001:
            b = p
            p = (b + (n / b)) / 2
            iteracoes += 1

    raiz_sqrt = math.sqrt(n)

    print(f"\nRaiz pelo método de Newton: {p:.10f}")
    print(f"Raiz pela função sqrt():   {raiz_sqrt:.10f}")
    print(f"Diferença entre n e p²: {abs(n - p ** 2):.10f}")
    print(f"Quantidade de iterações: {iteracoes}")
```
