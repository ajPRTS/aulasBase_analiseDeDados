```python
import json


def salvar_personagem():
    personagem = {
        "nome": input("Nome do personagem: "),
        "idade": int(input("Idade: ")),
        "classe": input("Classe: "),
        "nivel": int(input("Nível: ")),
        "forca": int(input("Força: ")),
        "defesa": int(input("Defesa: ")),
        "agilidade": int(input("Agilidade: "))
    }

    with open("personagem.json", "w", encoding="utf-8") as arquivo:
        json.dump(personagem, arquivo, indent=4, ensure_ascii=False)

    print("\nPersonagem salvo com sucesso!")


def ler_personagem():
    with open("personagem.json", "r", encoding="utf-8") as arquivo:
        personagem = json.load(arquivo)

    print("\n--- DADOS DO PERSONAGEM ---")

    for caracteristica, valor in personagem.items():
        print(f"{caracteristica.capitalize()}: {valor}")


salvar_personagem()
ler_personagem()
```
